```python
from langchain_core.runnables import RunnableLambda

def add_one(x):
    return x + 1

r_add = RunnableLambda(add_one)
print(r_add.invoke(3))
print(r_add.batch([1,2,3]))
```

```python
from langchain_core.runnables import RunnableLambda

r_add = RunnableLambda(lambda x: x+1)
r_double = RunnableLambda(lambda x: x*2)

chain_a = r_add | r_double
chain_b = r_double | r_add

print(chain_a.invoke(3))
print(chain_b.invoke(3))
print(type(chain_b))
```

```python
from langchain_core.runnables import (RunnableLambda, RunnableParallel, RunnablePassthrough)

analyzer = RunnableParallel(
    original = RunnablePassthrough(),
    upper = RunnableLambda(lambda x:x.upper()),
    length = RunnableLambda(lambda x: len(x))
)

print(analyzer.invoke("hello"))
```

```python
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

load_dotenv()   # .env 파일의 API키 불러오기

prompt = ChatPromptTemplate.from_template("{topic}에 대해서 부리부리대마왕처럼 설명해줘")

model = ChatOpenAI(model="gpt-4o-mini")
parser = StrOutputParser()

chain = prompt | model | parser

print(chain.invoke({"topic":"인공지능"}))
```

```python
results = chain.batch([{"topic":"블럭체인"},{"topic":"클라우드"}])

for chunck in chain.stream({"topic":"양자컴퓨터"}):
    print(chunck, end="", flush = True)
```


# 도전 과제

### 과제 1 (기초, API 키 불필요) — 문자열을 받아 양쪽 공백을 없애고, 대문자로 바꾼 뒤, 끝에 !를 붙이는 체인을 만드세요.

```python
from langchain_core.runnables import (RunnableLambda, RunnablePassthrough)

original = RunnablePassthrough()
strip = RunnableLambda(lambda x:x.strip())
upper = RunnableLambda(lambda x:x.upper())
add_exclamation = RunnableLambda(lambda x:x + "!")

chain_practice_1 =  original | strip | upper | add_exclamation

print(chain_practice_1.invoke("hello  "))
```

### 과제 2 (중급, API 키 불필요) — 숫자 리스트를 받아 합계, 평균, 최댓값을 한 번에 딕셔너리로 돌려주는 Runnable을 만드세요.

```python
from langchain_core.runnables import (RunnableLambda, RunnableParallel, RunnablePassthrough)

chain_practice_2 = RunnableParallel(
    sum = RunnableLambda(lambda x: sum(x)),
    avg = RunnableLambda(lambda x: sum(x) / len(x)),
    max_value = RunnableLambda(lambda x: max(x))
)

print(chain_practice_2.invoke([1, 2, 3, 4, 5]))
```

### 과제 3 (심화, API 키 필요) — 주제 하나를 넣으면 ① 한 줄 설명과 ② 관련 키워드 3개를 동시에 만들어 딕셔너리로 돌려주는 체인을 만드세요.

```python
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

load_dotenv()   # .env 파일의 API키 불러오기

prompt_1 = ChatPromptTemplate.from_template("{topic}에 대해서 한줄설명해줘")
prompt_2 = ChatPromptTemplate.from_template("{topic}에 대해서 관련 키워드 3개 만들어줘")

model = ChatOpenAI(model="gpt-4o-mini")
parser = StrOutputParser()

chain_practice_3 = RunnableParallel(
    prompt_1 = prompt_1 | model | parser,
    prompt_2 = prompt_2 | model | parser,
)


print(chain_practice_3.invoke({"topic":"인공지능"}))
```
